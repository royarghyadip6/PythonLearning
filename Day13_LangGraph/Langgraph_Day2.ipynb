# 📘 Schema Notes (TypedDict, dataclass, Pydantic)

---

## 1️⃣ What is a Schema?

A **schema** defines the **structure, types, and validation rules** of data.

### Why Schemas are Important

* Ensures **data correctness**
* Improves **readability & maintainability**
* Helps **validation & serialization**
* Widely used in **APIs, AI agents, RAG, LangChain, FastAPI**

---

## 2️⃣ TypedDict (from `typing`)

### 🔹 What is TypedDict?

`TypedDict` allows you to define **type hints for dictionary keys**.

📌 Used mainly for **static type checking**, not runtime validation.

---

### Example

```python
from typing import TypedDict

class User(TypedDict):
    id: int
    name: str
    email: str
```

Usage:

```python
user: User = {
    "id": 1,
    "name": "Arghya",
    "email": "a@x.com"
}
```

---

### Key Characteristics

| Feature            | TypedDict   |
| ------------------ | ----------- |
| Runtime validation | ❌ No        |
| Type checking      | ✅ Static    |
| Performance        | ✅ Very fast |
| Mutability         | Mutable     |
| JSON friendly      | ✅           |

---

### When to Use TypedDict

* Config objects
* LLM input/output structures
* Lightweight schemas
* When runtime validation is not needed

---

### Limitations

* No runtime enforcement
* No default values
* No complex validation logic

---

## 3️⃣ dataclass (from `dataclasses`)

### 🔹 What is dataclass?

`dataclass` is a Python feature to **store structured data as objects** with minimal boilerplate.

---

### Example

```python
from dataclasses import dataclass

@dataclass
class User:
    id: int
    name: str
    email: str
```

Usage:

```python
user = User(1, "Arghya", "a@x.com")
```

---

### With Default Values

```python
@dataclass
class Product:
    id: int
    name: str
    price: float = 0.0
```

---

### Key Characteristics

| Feature            | dataclass         |
| ------------------ | ----------------- |
| Runtime validation | ❌ No              |
| Type hints         | ✅                 |
| Mutability         | Mutable (default) |
| Object-oriented    | ✅                 |
| Performance        | ✅ Fast            |

---

### Advantages

* Cleaner code
* Built-in `__init__`, `__repr__`
* Good for **domain models**

---

### Limitations

* No automatic type validation
* No JSON validation
* Needs manual checks

---

## 4️⃣ Pydantic

### 🔹 What is Pydantic?

Pydantic provides **runtime data validation** using Python type hints.

📌 Widely used in **FastAPI, LangChain, LangGraph, RAG pipelines**

---

### Example

```python
from pydantic import BaseModel

class User(BaseModel):
    id: int
    name: str
    email: str
```

Usage:

```python
user = User(id="1", name="Arghya", email="a@x.com")
print(user.id)  # auto-converted to int
```

---

### Validation Example

```python
from pydantic import BaseModel, Field

class Product(BaseModel):
    id: int
    name: str
    price: float = Field(gt=0)
```

Invalid input:

```python
Product(id=1, name="Pen", price=-10)
```

➡ Raises **ValidationError**

---

### Key Characteristics

| Feature            | Pydantic          |
| ------------------ | ----------------- |
| Runtime validation | ✅                 |
| Type coercion      | ✅                 |
| JSON parsing       | ✅                 |
| Error handling     | ✅                 |
| Performance        | ⚠ Slightly slower |

---

## 5️⃣ Comparison Table

| Feature            | TypedDict    | dataclass     | Pydantic  |
| ------------------ | ------------ | ------------- | --------- |
| Runtime validation | ❌            | ❌             | ✅         |
| Static typing      | ✅            | ✅             | ✅         |
| JSON support       | ⚠ Limited    | ⚠ Limited     | ✅         |
| Defaults           | ❌            | ✅             | ✅         |
| Validation rules   | ❌            | ❌             | ✅         |
| Best for           | Light config | Domain models | APIs / AI |

---

## 6️⃣ When to Use What?

### ✅ Use TypedDict when:

* You need **lightweight typing**
* Performance matters
* Data is dictionary-based

### ✅ Use dataclass when:

* You need **structured objects**
* Business/domain models
* No runtime validation required

### ✅ Use Pydantic when:

* Validating external input
* API requests/responses
* LLM outputs
* RAG / Agent pipelines

---

## 7️⃣ Real-World Example (AI / RAG)

### LLM Output Schema (Pydantic)

```python
class Answer(BaseModel):
    answer: str
    sources: list[str]
```

Used to ensure **LLM responses are structured**.

---

## 8️⃣ Interview One-Liners

* **TypedDict** → “Static typing for dictionaries”
* **dataclass** → “Lightweight class for structured data”
* **Pydantic** → “Runtime data validation using type hints”

---

## 9️⃣ Summary

* Schemas define **shape + type + rules**
* TypedDict → type hints only
* dataclass → structured objects
* Pydantic → validation + serialization

---


````python
# =====================================
# Imports
# =====================================
from typing import TypedDict, List
from dataclasses import dataclass
from pydantic import BaseModel, Field, ValidationError


# =====================================
# 1️⃣ TypedDict Schema
# =====================================
class UserDict(TypedDict):
    """
    TypedDict schema:
    - Static type checking only
    - No runtime validation
    """
    id: int
    name: str
    email: str


def typed_dict_example():
    user: UserDict = {
        "id": 1,
        "name": "Arghya",
        "email": "user@test.com"
    }
    print("TypedDict User:", user)


# =====================================
# 2️⃣ dataclass Schema
# =====================================
@dataclass
class Product:
    """
    dataclass schema:
    - Structured object
    - No automatic validation
    """
    id: int
    name: str
    price: float = 0.0


def dataclass_example():
    product = Product(id=101, name="Pen", price=10.5)
    print("Dataclass Product:", product)


# =====================================
# 3️⃣ Pydantic Schema
# =====================================
class Order(BaseModel):
    """
    Pydantic schema:
    - Runtime validation
    - Type coercion
    - JSON serialization
    """
    order_id: int
    product_name: str
    quantity: int = Field(gt=0)
    price: float = Field(gt=0)

    @property
    def total_price(self) -> float:
        return self.quantity * self.price


def pydantic_example():
    try:
        order = Order(
            order_id="5001",          # auto-converted to int
            product_name="Notebook",
            quantity=2,
            price=50.0
        )
        print("Pydantic Order:", order)
        print("Total Price:", order.total_price)

    except ValidationError as e:
        print("Validation Error:", e)


# =====================================
# Run All Examples
# =====================================
if __name__ == "__main__":
    typed_dict_example()
    print("-" * 40)

    dataclass_example()
    print("-" * 40)

    pydantic_example()
````

# 📘 LangGraph State Concepts Notes

## 1️⃣ State (Message State)

In LangGraph, **state** represents the **shared data** that flows through the graph between nodes.

### Message State

* Most common state type
* Stores conversation history
* Typically a list of messages

Example:

```python
from langgraph.graph import MessagesState

state = {
    "messages": []
}
```

📌 `messages` is usually a list of:

* HumanMessage
* AIMessage
* ToolMessage

---

## 2️⃣ Reducer (State Reducer)

A **reducer** defines **how state values are merged** when multiple nodes update the same key.

### Why Reducers Are Needed

* Parallel node execution
* Branching paths
* Conflict resolution

Default reducer for `messages`:

* **Append messages**

---

### Example Reducer

```python
from typing import Annotated
from langgraph.graph import MessagesState
from langgraph.graph.message import add_messages

class State(MessagesState):
    messages: Annotated[list, add_messages]
```

📌 `add_messages` → appends new messages to the state

---

## 3️⃣ Branching

**Branching** allows the graph to follow **different paths based on state**.

### Why Branching Is Used

* Conditional logic
* Tool vs LLM decisions
* Multi-agent workflows

---

### Branch Example

```python
def route(state):
    if "tool" in state["messages"][-1].content.lower():
        return "tool_node"
    return "llm_node"
```

Graph routing:

```python
graph.add_conditional_edges(
    "router",
    route
)
```

---

## 4️⃣ Custom Reducer

A **custom reducer** lets you control **how state is combined**.

### Use Cases

* Deduplicate messages
* Limit history size
* Merge structured outputs

---

### Custom Reducer Example

```python
def last_message_only(old, new):
    return new[-1:]

class State(MessagesState):
    messages: Annotated[list, last_message_only]
```

📌 Keeps only the **latest message**

---

## 5️⃣ Message State

### What is Message State?

A structured state that:

* Tracks conversation history
* Supports tool calls
* Enables memory-like behavior

Example:

```python
state = {
    "messages": [
        HumanMessage(content="Hello"),
        AIMessage(content="Hi!")
    ]
}
```

---

## 6️⃣ Removing a Message from State

Sometimes you need to **remove or filter messages**.

### Use Case

* Prevent prompt bloat
* Remove sensitive info
* Clean tool chatter

---

### Message Removal Reducer

```python
def remove_tool_messages(old, new):
    messages = old + new
    return [m for m in messages if m.type != "tool"]
```

---

### Limit Message History

```python
def limit_history(old, new, limit=5):
    return (old + new)[-limit:]
```

---

## 7️⃣ State Flow Example (End-to-End)

```
User Input
   ↓
Message added to state
   ↓
Router checks last message
   ↓
Branch to LLM or Tool
   ↓
Reducer merges messages
   ↓
Final state returned
```

---

## 8️⃣ Why These Concepts Matter

| Concept        | Purpose               |
| -------------- | --------------------- |
| State          | Shared memory         |
| Reducer        | Conflict resolution   |
| Branching      | Conditional logic     |
| Custom Reducer | Advanced control      |
| Message State  | Conversation tracking |
| Remove Message | Memory optimization   |

---

## 9️⃣ Interview One-Liners

* **State** → “Shared data passed between graph nodes”
* **Reducer** → “Defines how state updates are merged”
* **Branching** → “Conditional execution paths”
* **Custom Reducer** → “Controls state combination logic”
* **Message State** → “Conversation memory”
* **Remove Message** → “Optimizes or cleans state”

---

## 10️⃣ Summary

* LangGraph uses **state-driven execution**
* Reducers are critical for correctness
* Branching enables agent reasoning
* Custom reducers enable memory control
* Message state powers conversational AI

---



````python
# ===============================
# Imports
# ===============================
from typing import Annotated
from langgraph.graph import StateGraph, MessagesState
from langgraph.graph.message import add_messages
from langchain_core.messages import HumanMessage, AIMessage, ToolMessage

# ===============================
# Custom Reducer
# ===============================
def remove_tool_and_limit_history(old_messages, new_messages):
    """
    1. Merge old + new messages
    2. Remove tool messages
    3. Keep only last 5 messages
    """
    combined = old_messages + new_messages

    # Remove tool messages
    filtered = [m for m in combined if m.type != "tool"]

    # Keep last 5 messages only
    return filtered[-5:]


# ===============================
# State Definition
# ===============================
class State(MessagesState):
    messages: Annotated[list, remove_tool_and_limit_history]


# ===============================
# Graph Node
# ===============================
def chatbot_node(state: State):
    """
    Simple chatbot node that replies to user input
    """
    last_message = state["messages"][-1]

    reply = f"Echo: {last_message.content}"

    return {
        "messages": [
            AIMessage(content=reply)
        ]
    }


# ===============================
# Build Graph
# ===============================
graph = StateGraph(State)

graph.add_node("chatbot", chatbot_node)
graph.set_entry_point("chatbot")
graph.set_finish_point("chatbot")

app = graph.compile()


# ===============================
# Run the Graph
# ===============================
if __name__ == "__main__":
    result = app.invoke(
        {
            "messages": [
                HumanMessage(content="Hello"),
                ToolMessage(content="tool output", tool_name="dummy"),
                HumanMessage(content="How are you?")
            ]
        }
    )

    print("Final Messages in State:\n")
    for msg in result["messages"]:
        print(f"{msg.type.upper()}: {msg.content}")
````

# 📘 Schema Concepts Notes

---

## 1️⃣ Multiple Schema

### 🔹 What is Multiple Schema?

Using **more than one schema** in an application, where each schema has a **specific responsibility**.

### Why Multiple Schemas Are Used

* Clear separation of concerns
* Strong validation boundaries
* Better security
* Easier maintenance

### Common Types in Real Applications

* Input Schema
* Internal (Domain) Schema
* Output Schema
* Overall / Aggregated Schema

---

### Example (E-commerce / AI App)

```
User Request  → Input Schema
Processing    → Internal Schema
Response      → Output Schema
```

---

## 2️⃣ Input Schema

### 🔹 Definition

An **Input Schema** defines the **exact structure of incoming data**.

### Purpose

* Validate user/API/tool input
* Reject invalid or malicious data
* Enforce required fields

### Characteristics

* Strict validation
* Type enforcement
* Value constraints

### Example Use Cases

* API request body
* Tool input in LangGraph
* LLM prompt parameters

---

## 3️⃣ Output Schema

### 🔹 Definition

An **Output Schema** defines the **structure of data returned** by the system.

### Purpose

* Control exposed fields
* Format response consistently
* Ensure predictable output

### Characteristics

* Clean and minimal
* Serialization-ready (JSON)
* No sensitive data

### Example Use Cases

* API responses
* LLM structured outputs
* RAG answers

---

## 4️⃣ Overall Schema

### 🔹 Definition

An **Overall Schema** represents the **complete data flow or final response**, often combining multiple outputs.

### Purpose

* High-level response structure
* Wraps metadata + result
* Used in APIs and AI systems

### Characteristics

* Aggregates multiple schemas
* Used at system boundaries
* Improves clarity

---

### Example Overall Schema Structure

```json
{
  "status": "success",
  "data": { ...output schema... },
  "meta": { ... }
}
```

---

## 5️⃣ Comparison Summary

| Schema Type     | Responsibility           |
| --------------- | ------------------------ |
| Input Schema    | Validate incoming data   |
| Internal Schema | Business logic           |
| Output Schema   | Control response shape   |
| Overall Schema  | Aggregate final response |

---

## 6️⃣ Interview One-Liners

* **Multiple Schema** → “Different schemas for different stages of data”
* **Input Schema** → “Validates incoming data”
* **Output Schema** → “Controls returned data”
* **Overall Schema** → “Wraps complete system response”

---

## 🧠 End-to-End Flow

```
Raw Input
   ↓ Input Schema
Validated Data
   ↓ Internal Schema
Business Logic
   ↓ Output Schema
Clean Result
   ↓ Overall Schema
Final Response
```

---


````python
# =====================================
# Imports
# =====================================
from pydantic import BaseModel, Field
from dataclasses import dataclass
from typing import Optional


# =====================================
# 1️⃣ Input Schema
# =====================================
class OrderInput(BaseModel):
    """
    Validates incoming user / API / tool data
    """
    product_name: str = Field(min_length=2)
    price: float = Field(gt=0)
    quantity: int = Field(gt=0)


# =====================================
# 2️⃣ Internal (Domain) Schema
# =====================================
@dataclass
class OrderInternal:
    """
    Used for business logic processing
    """
    product_name: str
    price: float
    quantity: int

    def total_amount(self) -> float:
        return self.price * self.quantity


# =====================================
# 3️⃣ Output Schema
# =====================================
class OrderOutput(BaseModel):
    """
    Defines the response data structure
    """
    product_name: str
    total_amount: float


# =====================================
# 4️⃣ Overall Schema
# =====================================
class APIResponse(BaseModel):
    """
    Wraps the complete response
    """
    status: str
    data: OrderOutput
    message: Optional[str] = None


# =====================================
# Application Flow
# =====================================
if __name__ == "__main__":
    # Simulated input
    raw_input = {
        "product_name": "Mobile",
        "price": 15000,
        "quantity": 2
    }

    # Step 1: Input validation
    validated_input = OrderInput(**raw_input)

    # Step 2: Internal processing
    internal_order = OrderInternal(
        product_name=validated_input.product_name,
        price=validated_input.price,
        quantity=validated_input.quantity
    )

    # Step 3: Prepare output
    output = OrderOutput(
        product_name=internal_order.product_name,
        total_amount=internal_order.total_amount()
    )

    # Step 4: Overall response
    response = APIResponse(
        status="success",
        data=output,
        message="Order processed successfully"
    )

    print(response.json(indent=2))
````

# 🧠 Memory and Threads in LangGraph

---

## 1️⃣ Memory in LangGraph

### 🔹 What is Memory?

In LangGraph, **memory is simply state**.

Anything stored in the **graph state** and passed between nodes is considered **memory**.

📌 There is **no separate “memory object”**—memory is:

* State
* Messages
* Structured data

---

### 🔹 What Can Be Memory?

* Conversation messages
* Tool outputs
* Intermediate reasoning
* User preferences
* RAG retrieval results

---

### 🔹 Memory Characteristics

| Property     | Description                |
| ------------ | -------------------------- |
| Scoped       | Per execution / per thread |
| Mutable      | Updated by nodes           |
| Reducible    | Controlled by reducers     |
| Serializable | Can be persisted           |

---

## 2️⃣ Types of Memory in LangGraph

### 🔹 1. Short-Term Memory

* Lives within a thread execution
* Usually message history
* Token-limited

📌 Implemented via:

* `MessagesState`
* Sliding window reducers

---

### 🔹 2. Long-Term Memory

* Persists across executions
* External storage

📌 Examples:

* Vector DB
* SQL / NoSQL DB
* Files

📌 LangGraph does **not force** a storage type—you manage persistence.

---

### 🔹 3. Working Memory

* Temporary, step-level memory
* Used for reasoning

📌 Examples:

* Tool results
* Partial plans
* Scratchpad values

---

## 3️⃣ Reducers = Memory Control

Reducers decide:

* What to keep
* What to discard
* How memory merges

📌 Reducers are the **memory policy layer**.

---

## 4️⃣ Thread in LangGraph

### 🔹 What is a Thread?

A **thread** represents **one independent execution context** of a graph.

Each thread has:

* Its own state (memory)
* Its own execution path
* No shared memory by default

📌 Think of a thread as:

> “One conversation or one workflow instance”

---

## 5️⃣ Why Threads Are Important

* Multi-user systems
* Parallel executions
* Isolated memory
* Stateful agents

Without threads:

* Memory collisions
* Incorrect responses
* Data leakage

---

## 6️⃣ Thread vs Memory (Key Difference)

| Concept  | Memory             | Thread            |
| -------- | ------------------ | ----------------- |
| What     | Stored data        | Execution context |
| Scope    | Inside a thread    | Contains memory   |
| Purpose  | Remember info      | Isolate workflows |
| Lifetime | Reducer-controlled | Per invocation    |

📌 **Thread owns memory**, not the other way around.

---

## 7️⃣ Thread Lifecycle

```
Thread starts
   ↓
Initial state injected
   ↓
Nodes execute
   ↓
State (memory) updated
   ↓
Thread ends or persists
```

---

## 8️⃣ Memory Across Threads

### ❌ By Default

* Threads are isolated
* Memory is NOT shared

### ✅ If Needed

* Persist memory externally
* Reload into new thread

📌 Example:

* Load user profile from DB
* Inject into thread state

---

## 9️⃣ Threads in Real Systems

### Example Use Cases

| Use Case       | Thread     |
| -------------- | ---------- |
| Chat session   | One thread |
| API request    | One thread |
| Agent task     | One thread |
| Background job | One thread |

---

## 🔟 Memory + Thread Together (Mental Model)

```
User Session
   ↓
Thread Created
   ↓
State Initialized (Memory)
   ↓
Graph Executes
   ↓
State Mutates
   ↓
Thread Ends / Saves State
```

---

## 11️⃣ Common Misconceptions

❌ “Memory is global”
✔ Memory is **thread-scoped**

❌ “Threads share memory”
✔ Threads are **isolated**

❌ “LangGraph stores memory automatically”
✔ You control persistence

---

## 12️⃣ Interview One-Liners

* **Memory** → “State that persists during graph execution”
* **Thread** → “Isolated execution context with its own memory”
* **Reducer** → “Memory management policy”
* **Thread safety** → “Guaranteed by state isolation”

---

## 13️⃣ Summary

* Memory = state
* Reducers control memory behavior
* Threads isolate memory
* Each thread = one workflow instance
* Persistence is optional and external

---



````python
# =========================================
# Imports
# =========================================
from typing import Annotated
from langgraph.graph import StateGraph, MessagesState
from langgraph.graph.message import add_messages
from langchain_core.messages import HumanMessage, AIMessage


# =========================================
# Memory Reducer (Sliding Window)
# Controls short-term memory
# =========================================
def memory_reducer(old_messages, new_messages, limit=5):
    """
    Merges old + new messages
    Keeps only last N messages
    """
    return (old_messages + new_messages)[-limit:]


# =========================================
# Thread State (Memory lives here)
# =========================================
class ThreadState(MessagesState):
    messages: Annotated[list, memory_reducer]


# =========================================
# Agent Node
# =========================================
def agent_node(state: ThreadState):
    """
    Agent uses thread memory (state)
    """
    last_message = state["messages"][-1].content

    response = f"I remember this from our thread: '{last_message}'"

    return {
        "messages": [
            AIMessage(content=response)
        ]
    }


# =========================================
# Build Graph
# =========================================
graph = StateGraph(ThreadState)
graph.add_node("agent", agent_node)
graph.set_entry_point("agent")
graph.set_finish_point("agent")

app = graph.compile()


# =========================================
# Run Two Separate Threads
# =========================================
if __name__ == "__main__":

    # -------- Thread 1 --------
    thread_1_state = {
        "messages": [
            HumanMessage(content="Hi"),
            HumanMessage(content="My name is Arghya")
        ]
    }

    result_1 = app.invoke(thread_1_state)

    print("Thread 1 Memory:")
    for msg in result_1["messages"]:
        print(f"{msg.type.upper()}: {msg.content}")

    print("\n" + "-" * 40 + "\n")

    # -------- Thread 2 --------
    thread_2_state = {
        "messages": [
            HumanMessage(content="Hello"),
            HumanMessage(content="I like LangGraph")
        ]
    }

    result_2 = app.invoke(thread_2_state)

    print("Thread 2 Memory:")
    for msg in result_2["messages"]:
        print(f"{msg.type.upper()}: {msg.content}")
````
