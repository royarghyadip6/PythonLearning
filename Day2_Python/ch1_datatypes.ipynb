## Mutable and Immutable Object
* For immutable objects the referance is being changes so the id will be different but mutable object have same id.
* Immutable object : Numbers, Strings, Tuples, Frozen Sets
* Mutable object : Lists, Dictionaries, Sets, Byte Arrays

```python
# Immutable object
SUGAR = 2
print(f"Initial Sugar : {SUGAR}")
print(f"Identity of 2 is : {id(2)} and SUGAR is {id(SUGAR)}")
SUGAR = 12
print(f"Second Sugar : {SUGAR}")
print(f"Identity of 12 is : {id(12)} and SUGAR is {id(SUGAR)}")

# Mutable object
SPICES = set()
print(f"Initial SPICES : {SPICES}")
print(f"Identity of SPICES is : {id(SPICES)}")
SPICES.add("peeper")
print(f"Initial SPICES : {SPICES}")
print(f"Identity of SPICES is : {id(SPICES)}")
```


#### Boolean Data Type (bool)

Boolean is a built-in data type in Python that has two constant values:
* **True**: The Boolean value 'True' represents 1 and is returned when the expression or object are considered truthful.
* **False**: The Boolean value 'False' represents 0 and is returned when the expression or object turns out to be false.

We can also evaluate non-Boolean objects in a Boolean context and identify them as either True or False. A class called 'bool' is used to store the Boolean objects in Python.

**Note: Python is case-sensitive; therefore, it will return error if the Boolean values (True and False) is passed in lowercase (e.g., true, false).**


```python
# Boolean Operations in Python
x = True
y = False
print("Boolean Operations:")
print("AND:", x and y)     # False
print("OR:", x or y)       # True
print("NOT x:", not x)     # False
```

```python
# Comparison + Boolean Result
p = 5
q = 10
print("Comparison Operations:")
print("p > q:", p > q)        # False
print("p < q:", p < q)        # True
print("p == q:", p == q)      # False
print("p != q:", p != q)      # True
print("p <= q:", p <= q)      # True
print("p >= q:", p >= q)      # False
```


#### String (str)

A String is a sequence of characters enclosed in single (') or double (") quotes. Similar to Tuples, strings are also immutable in nature. In Python, the strings are represented using the 'str' class.

```python
s = "hello world"

# ===== CASE =====
print(s.lower())        # convert to lower → "hello world"
print(s.upper())        # convert to upper → "HELLO WORLD"
print(s.title())        # title case words → "Hello World"
print(s.capitalize())   # first letter upper → "Hello world"
print(s.swapcase())     # swap upper/lower → "HELLO WORLD"
print(s.casefold())     # strong lower compare → "hello world"

# ===== FIND / SEARCH =====
print(s.find("o"))      # first index find → 4
print(s.rfind("o"))     # last index find → 7
print(s.index("o"))     # find else error → 4
print("lo" in s)        # substring exists → True

# ===== COUNT =====
print(s.count("o"))     # count occurrences → 2

# ===== CHECK BOOLEAN =====
print("abc".isalpha())      # only letters check → True
print("123".isdigit())      # only digits check → True
print("abc123".isalnum())   # letters + digits → True
print("   ".isspace())      # only spaces check → True
print("hello".islower())    # all lowercase check → True
print("HELLO".isupper())    # all uppercase check → True
print("Hello World".istitle())  # title format check → True
print("abc".isascii())      # ascii chars check → True
print("123".isdecimal())    # decimal numbers check → True
print("123".isnumeric())    # numeric value check → True
print("var_1".isidentifier())   # valid variable name → True
print("hello".isprintable())    # printable chars check → True

# ===== START / END =====
print(s.startswith("he"))   # starts with check → True
print(s.endswith("ld"))     # ends with check → True

# ===== REPLACE / STRIP =====
print(s.replace("world","python"))   # replace substring → "hello python"
print("  hello  ".strip())           # remove both sides → "hello"
print("  hello  ".lstrip())          # remove left spaces → "hello  "
print("  hello  ".rstrip())          # remove right spaces → "  hello"

# ===== PREFIX / SUFFIX =====
print("hello".removeprefix("he"))   # remove start text → "llo"
print("hello".removesuffix("lo"))   # remove end text → "hel"

# ===== SPLIT / JOIN =====
print(s.split())            # split by space → ["hello","world"]
print(s.split("o"))         # split by char → ["hell"," w","rld"]
print("line1\nline2".splitlines())  # split lines → ["line1","line2"]
print("-".join(["a","b","c"]))      # join list string → "a-b-c"

# ===== PARTITION =====
print(s.partition(" "))     # split first match → ("hello"," ","world")
print(s.rpartition(" "))    # split last match → ("hello"," ","world")

# ===== ALIGN =====
print(s.center(15))         # center with spaces → "  hello world  "
print(s.ljust(15))          # left align text → "hello world    "
print(s.rjust(15))          # right align text → "    hello world"
print("42".zfill(5) )       # pad with zeros → "00042"

# ===== FORMAT =====
name="Arghya"; age=25
print(f"{name} {age}")                  # fstring format text → "Arghya 25"
print("{} {}".format(name,age))         # format placeholders → "Arghya 25"
print("{name} {age}".format(name=name,age=age))  # named format → "Arghya 25"

# ===== ENCODE / DECODE =====
print("hello".encode())     # string to bytes → b'hello'
print(b"hello".decode())    # bytes to string → "hello"

# ===== TRANSLATE =====
table=str.maketrans({"h":"H"})
print("hello".translate(table))   # character mapping replace → "Hello"

# ===== EXPAND TABS =====
print("hi\tthere".expandtabs(4))  # tab to spaces → "hi  there"

# ===== INDEX / SLICING =====
print(s[0])        # first character access → "h"
print(s[-1])       # last character access → "d"
print(s[0:5])      # substring slice range → "hello"
print(s[::-1])     # reverse string slice → "dlrow olleh"

# ===== LENGTH =====
print(len(s))      # string length count → 11

# ===== CONCAT / REPEAT =====
print("a"+"b")     # concatenate strings → "ab"
print("a"*3)       # repeat string times → "aaa"

# ===== STRING CONSTANTS =====
import string
print(string.ascii_lowercase)   # all lowercase letters → "abcdefghijklmnopqrstuvwxyz"
print(string.ascii_uppercase)   # all uppercase letters → "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
print(string.digits)            # numeric characters → "0123456789"
print(string.punctuation)       # punctuation symbols → !"#$...
print(string.whitespace)        # whitespace characters → space tab newline
```


#### Tuples (tuple)

Tuples are the ordered sequence of data elements same as the lists. The only difference between the two is that Tuples are immutable in nature. This means that we cannot modify the elements of the tuple once created.

In Python, we use parentheses '()' in order to store elements in tuple.

```python
# TUPLE EXAMPLES
t = (10, 20, 30, 20)
print("Tuple:", t)
# Access
print("First element:", t[0])
# Count & Index
print("Count of 20:", t.count(20))
print("Index of 30:", t.index(30))
# Tuple is immutable (cannot change)
# t[0] = 100  ❌ ERROR
```


#### Lists (list)

Similar to array in other programming languages, List in Python is an ordered collection of data elements. It can simultaneously store elements of different data types separately by commas. Lists are mutable in nature, meaning that we can modify its elements anytime.

```python
# LIST EXAMPLES
lst = [1, 2, 3]
print("List:", lst)

# Add elements
lst.append(4)
lst.extend([5, 6])
print("After append & extend:", lst)

# Insert
lst.insert(1, 100)
print("After insert:", lst)

# Remove
lst.remove(3)
print("After remove:", lst)

# Pop
lst.pop()
print("After pop:", lst)

# Sort & Reverse
lst.sort()
lst.reverse()
print("Sorted & Reversed:", lst)
```

```python
# BYTEARRAY EXAMPLES
b = bytearray([65, 66, 67])
print("Bytearray:", b)
print("As string:", b.decode())

# Modify bytearray (mutable)
b[0] = 90
print("Modified bytearray:", b)
print("As string:", b.decode())

# Append
b.append(68)
print("After append:", b.decode())
```


#### set

A set in Python is said to be an unordered, mutable and unindexed collection of unique data elements. Sets are useful in storing distinct values and perform various mathematical operations like union, intersection, and difference.

In order to create a set in Python, we need to use {} or the set() constructor. Let us take an example illustrating the ways of initializing sets in Python.

```python
# -------- SET --------
# Create a set
s = {1, 2, 3, 3, 4}
print("Set:", s)   # duplicates removed
set1 = set(["apple","banana"])
print(set1)
# Add element
s.add(5)
print("After add:", s)

# Remove element
s.remove(2)
print("After remove:", s)

# Set operations
a = {1, 2, 3}
b = {3, 4, 5}

print("Union:", a | b)
print("Intersection:", a & b)
print("Difference:", a - b)
print("Symmetric Difference:", a ^ b)

# Membership
print("Is 3 in set?", 3 in a)
```


#### Frozenset

In Python, a function called frozenset() is used to return a frozenset object. This object is an immutable, unordered data type. A frozenset consists of unique items, making it similar to the set data type; however, it is unchangeable like tuples.

Let us look at an example illustrating the way of initializing a frozenset in Python.

```python
# -------- FROZEN SET --------
# Create frozenset
fs = frozenset([1, 2, 3, 3])
print("FrozenSet:", fs)

# Set operations work
fs1 = frozenset([1, 2, 3])
fs2 = frozenset([3, 4])

print("Union:", fs1 | fs2)
print("Intersection:", fs1 & fs2)
print("Difference:", fs1 - fs2)

# Immutable (will cause error if uncommented)
# fs.add(4)

print("FrozenSet is immutable")
```

**Counter**
Counts frequency of elements in an iterable. Returns a dict-like object with element counts.

**defaultdict**
Dictionary that automatically assigns a default value for missing keys. Prevents KeyError.

**OrderedDict**
Dictionary that preserves insertion order and allows reordering of keys. Mostly legacy now.

**deque**
Double-ended queue with fast append and pop from both ends. Ideal for queues and stacks.

**namedtuple**
Tuple with named fields for better readability. Immutable and memory-efficient.

**ChainMap**
Combines multiple dictionaries into a single view. Searches keys sequentially.

```python
# advanced_collections_examples.py

from collections import (
    Counter,
    defaultdict,
    OrderedDict,
    deque,
    namedtuple,
    ChainMap
)

print("=== Counter ===")
data = ["apple", "banana", "apple", "orange", "banana", "apple"]
counter = Counter(data)
print(counter)                  # Count of elements
print(counter["apple"])         # Frequency of apple

print("\n=== defaultdict ===")
dd = defaultdict(int)           # default value = 0
dd["a"] += 1
dd["b"] += 2
print(dict(dd))

print("\n=== OrderedDict ===")
od = OrderedDict()
od["name"] = "Arghya"
od["role"] = "Developer"
od["company"] = "Nokia"
print(od)

print("\n=== deque (Double-ended Queue) ===")
dq = deque([1, 2, 3])
dq.append(4)        # add right
dq.appendleft(0)    # add left
dq.pop()            # remove right
dq.popleft()        # remove left
print(dq)

print("\n=== namedtuple ===")
Person = namedtuple("Person", ["name", "age", "city"])
p1 = Person("Arghya", 28, "Kolkata")
print(p1)
print(p1.name, p1.age)

print("\n=== ChainMap ===")
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 20, "c": 30}
cm = ChainMap(dict1, dict2)
print(cm)
print(cm["b"])     # Takes value from first dict

print("\n=== Summary ===")
print("Counter      -> Counts elements")
print("defaultdict  -> Auto default values")
print("OrderedDict  -> Maintains insertion order")
print("deque        -> Fast append/pop from both ends")
print("namedtuple   -> Tuple with named fields")
print("ChainMap     -> Combines multiple dicts")
```
