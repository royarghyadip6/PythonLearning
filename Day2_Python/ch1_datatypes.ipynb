# **Datatypes in python**

## Mutable and Immutable Object
For immutable objects the referance is being changes so the id will be different but mutable object have same id.
Immutable object : Numbers, Strings, Tuples, Frozen Sets
Mutable object : Lists, Dictionaries, Sets, Byte Arrays

```python
# Immutable object
SUGAR = 2
print(f"Initial Sugar : {SUGAR}")
print(f"Identity of 2 is : {id(2)} and SUGAR is {id(SUGAR)}")
SUGAR = 12
print(f"Second Sugar : {SUGAR}")
print(f"Identity of 12 is : {id(12)} and SUGAR is {id(SUGAR)}")

# Mutable object
SPICES = set()
print(f"Initial SPICES : {SPICES}")
print(f"Identity of SPICES is : {id(SPICES)}")
SPICES.add("peeper")
print(f"Initial SPICES : {SPICES}")
print(f"Identity of SPICES is : {id(SPICES)}")
```

```python
# ===============================
# Arithmetic Operations in Python
# ===============================

a = 10
b = 3

print("Arithmetic Operations:")
print("Addition:", a + b)          # 13
print("Subtraction:", a - b)       # 7
print("Multiplication:", a * b)    # 30
print("Division:", a / b)          # 3.333...
print("Floor Division:", a // b)   # 3
print("Modulus:", a % b)           # 1
print("Exponent:", a ** b)         # 1000
```

```python
# Boolean Operations in Python
x = True
y = False
print("Boolean Operations:")
print("AND:", x and y)     # False
print("OR:", x or y)       # True
print("NOT x:", not x)     # False
```

```python
# Comparison + Boolean Result
p = 5
q = 10
print("Comparison Operations:")
print("p > q:", p > q)        # False
print("p < q:", p < q)        # True
print("p == q:", p == q)      # False
print("p != q:", p != q)      # True
print("p <= q:", p <= q)      # True
print("p >= q:", p >= q)      # False
```

```python
# String creation
s1 = "Hello"
s2 = "World"
# Concatenation
print("Concat:", s1 + " " + s2)
# Repetition
print("Repeat:", s1 * 3)
# Length
print("Length:", len(s1))
# Indexing
print("First char:", s1[0])
# Slicing
print("Slice:", s1[1:4])
# Membership
print("'e' in s1:", 'e' in s1)
# Comparison
print("Compare:", s1 == "Hello")
# String methods
text = "  Python Programming  "
print("Upper:", text.upper())
print("Lower:", text.lower())
print("Strip:", text.strip())
print("Replace:", text.replace("Python", "Java"))
print("Split:", text.split())
print("Startswith:", text.startswith("  Py"))
print("Endswith:", text.endswith("ing  "))
print("Find:", text.find("Prog"))
print("Count:", text.count("o"))
```

```python
# TUPLE EXAMPLES
t = (10, 20, 30, 20)
print("Tuple:", t)
# Access
print("First element:", t[0])
# Count & Index
print("Count of 20:", t.count(20))
print("Index of 30:", t.index(30))
# Tuple is immutable (cannot change)
# t[0] = 100  ❌ ERROR
```

```python
# LIST EXAMPLES
lst = [1, 2, 3]
print("List:", lst)

# Add elements
lst.append(4)
lst.extend([5, 6])
print("After append & extend:", lst)

# Insert
lst.insert(1, 100)
print("After insert:", lst)

# Remove
lst.remove(3)
print("After remove:", lst)

# Pop
lst.pop()
print("After pop:", lst)

# Sort & Reverse
lst.sort()
lst.reverse()
print("Sorted & Reversed:", lst)
```

```python
# BYTEARRAY EXAMPLES
b = bytearray([65, 66, 67])
print("Bytearray:", b)
print("As string:", b.decode())

# Modify bytearray (mutable)
b[0] = 90
print("Modified bytearray:", b)
print("As string:", b.decode())

# Append
b.append(68)
print("After append:", b.decode())
```

```python
# -------- SET --------
# Create a set
s = {1, 2, 3, 3, 4}
print("Set:", s)   # duplicates removed

# Add element
s.add(5)
print("After add:", s)

# Remove element
s.remove(2)
print("After remove:", s)

# Set operations
a = {1, 2, 3}
b = {3, 4, 5}

print("Union:", a | b)
print("Intersection:", a & b)
print("Difference:", a - b)
print("Symmetric Difference:", a ^ b)

# Membership
print("Is 3 in set?", 3 in a)
```

```python
# -------- FROZEN SET --------
# Create frozenset
fs = frozenset([1, 2, 3, 3])
print("FrozenSet:", fs)

# Set operations work
fs1 = frozenset([1, 2, 3])
fs2 = frozenset([3, 4])

print("Union:", fs1 | fs2)
print("Intersection:", fs1 & fs2)
print("Difference:", fs1 - fs2)

# Immutable (will cause error if uncommented)
# fs.add(4)

print("FrozenSet is immutable")
```

```python
# advanced_collections_examples.py

from collections import (
    Counter,
    defaultdict,
    OrderedDict,
    deque,
    namedtuple,
    ChainMap
)

print("=== Counter ===")
data = ["apple", "banana", "apple", "orange", "banana", "apple"]
counter = Counter(data)
print(counter)                  # Count of elements
print(counter["apple"])         # Frequency of apple

print("\n=== defaultdict ===")
dd = defaultdict(int)           # default value = 0
dd["a"] += 1
dd["b"] += 2
print(dict(dd))

print("\n=== OrderedDict ===")
od = OrderedDict()
od["name"] = "Arghya"
od["role"] = "Developer"
od["company"] = "Nokia"
print(od)

print("\n=== deque (Double-ended Queue) ===")
dq = deque([1, 2, 3])
dq.append(4)        # add right
dq.appendleft(0)    # add left
dq.pop()            # remove right
dq.popleft()        # remove left
print(dq)

print("\n=== namedtuple ===")
Person = namedtuple("Person", ["name", "age", "city"])
p1 = Person("Arghya", 28, "Kolkata")
print(p1)
print(p1.name, p1.age)

print("\n=== ChainMap ===")
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 20, "c": 30}
cm = ChainMap(dict1, dict2)
print(cm)
print(cm["b"])     # Takes value from first dict

print("\n=== Summary ===")
print("Counter      -> Counts elements")
print("defaultdict  -> Auto default values")
print("OrderedDict  -> Maintains insertion order")
print("deque        -> Fast append/pop from both ends")
print("namedtuple   -> Tuple with named fields")
print("ChainMap     -> Combines multiple dicts")
```
