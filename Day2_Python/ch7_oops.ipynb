# **Object Oriented programming in python**

#### Class and Object Namespace
**Class** : A class is a collection of objects. Classes are blueprints for creating objects. A class defines a set of attributes and methods that the created objects (instances) can have.
* Classes are created by keyword class.
* Attributes are the variables that belong to a class.
* Attributes are always public and can be accessed using the dot (.) operator. Example: Myclass.Myattribute

**Object** : An Object is an instance of a Class. It represents a specific implementation of the class and holds its own data.
An object consists of:
* State: It is represented by the attributes and reflects the properties of an object.
* Behavior: It is represented by the methods of an object and reflects the response of an object to other objects.
* Identity: It gives a unique name to an object and enables one object to interact with other objects.

**Namespace** : Each object has its own entity/features which is called namespace. It does not effect other objects and original class also.

```python
class Chai:
    origin = "India"    # Variable inside class is called properties
    pass

print(type(Chai))

# Add property
Chai.is_hot = True
print(Chai.is_hot)

# Creating object form class Chai
masalaChai = Chai()
print("Masala chai origin: ",masalaChai.origin)
print("Masala chai origin: ",masalaChai.is_hot)
```

#### Attribute Shadowing
Variable inside the class is called attribute.

```python
class Chai:
    temperature = "Hot"
    strength = "Strong"

print("Class > temp :",Chai.temperature," Strength:",Chai.strength)

chai = Chai()
print("Object > temp :",chai.temperature," Strength:",chai.strength)
chai.strength = "Mild"
chai.cupSize = "Small"
print("Class Before Delete >  Strength:",Chai.strength)
print("Object Before Delete > Strength:",chai.strength," Cup Size:",chai.cupSize)
del chai.strength
del chai.cupSize
print("After Delete Class >  Strength:",Chai.strength)
print("After Delete Object var > Strength:",chai.strength)  # After deleting variable also the data is coming from class attribute. The value will be retrieved from default fallback.
print("After Delete Object var > Cup Size:",chai.cupSize)   # Variable is created outside class and after deletion of it's throwing error.
```

#### Self Argument
* If you call a method via an object → DON’T pass self
* If you call via class → YOU must pass the object

```python
class Dog:
    bread = "Desi"
    def describe(self):
        return f"This is {self.bread} dog."

dog1 = Dog()
dog1.bread = "abc"
print(dog1.describe(dog1))  # Calling method via object

dog2 = Dog()
dog2.bread = "German Shaperd"
print(Dog.describe(dog2))   # Calling method via class
```

#### Constructor and init
__init__ is a inbuild keyword to initilize a constructor

```python
class ChaiCup:
    def __init__(self, size, type_):
        self.size = size
        self.type = type_

    def summary(self):
        return f"{self.type} cup is of {self.size}ml"

cup = ChaiCup(20,"Normal")
print(cup.summary())
```

#### Inheritance and Composition
For inheritance syntax is to pass parent class to the child class.

```python
class BaseChai:
    def __init__(self, type_):
        self.type = type_

    def prepare(self):
        print(f"Preparing {self.type} chai....")

"""
Inheritance
This MasalaChai inherits BaseChai. Syntax is to pass parent class to the child class.
"""
class MasalaChai(BaseChai):
    def add_spices(self):
        print("Adding cardamom, ginger, cloves.")

"""
Composition
For composition syntax is to
"""
class ChaiShop:
    chai_cls = BaseChai

    def __init__(self):
        self.chai = self.chai_cls("Regular")

    def serve(self):
        print(f"Serving {self.chai.type} chai in the shop")
        self.chai.prepare()

"""
FancyChaiShop is inheriting ChaiShop
FancyChaiShop is compositing MasalaChai
"""
class FancyChaiShop(ChaiShop):
    chai_cls = MasalaChai

shop = ChaiShop()
fancy = FancyChaiShop()
shop.serve()
fancy.serve()
fancy.chai.add_spices()
```

#### Polymorphism
**_Compile-time polymorphism_** : In languages like Java or C++, compile-time polymorphism is achieved through method overloading but it's not directly supported in Python. In Python:
* True compile-time polymorphism is not supported.
* Instead, Python mimics it using default arguments or *args/**kwargs.
* Operator overloading can also be seen as part of polymorphism, though it is implemented at runtime in Python.

**_Run-Time Polymorphism_** : Run-Time Polymorphism is determined during the execution of the program. It covers multiple forms in Python:
* Method Overriding: A subclass redefines a method from its parent class.
* Duck Typing: If an object implements the required method, it works regardless of its type.
* Operator Overloading: Special methods (__add__, __sub__, etc.) redefine how operators behave for user-defined objects.

```python
# Compile time polymorphism
class Calculator:
    def add(self, *args):
        return sum(args)

calc = Calculator()
print(calc.add(5, 10))       # Two arguments
print(calc.add(5, 10, 15))   # Three arguments
print(calc.add(1, 2, 3, 4))  # Any number of arguments
```

```python
# Method Overriding
# We start with a base class and then a subclass that "overrides" the speak method.
class Animal:
    def speak(self):
        return f'Animal speaks.'

class Dog(Animal):
    def speak(self):
        return f'Barks'

print(Dog().speak())

# Duck Typing
class Cat(Animal):
    def speak(self):
        return f'Meaowww'

# This function works for both Dog and Cat because they both have a 'speak' method.
def make_animal_speak(abc):
    return abc.speak()

cat = Cat()
print(make_animal_speak(cat))
print(make_animal_speak(Dog()))


# 3 Operator Overloading
# We create a simple class that customizes the '+' operator.
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        # This special method defines the behavior of the '+' operator.
        return Vector(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(2, 3)
v2 = Vector(4, 5)
v3 = v1 + v2

print(v3)
```
