```python
# printing numbers using for loop
for i in range(1,7): #Here we have used for loop
    print (i)
```

```python
# given list
cars = ["Tata", "Honda", "Mahindra", "Suzuki", "BMW"]
# using for loop to iterate each element from the list
for brands in cars:
  print(brands)
```

```python
# given tuple
tuple_1 = ("Python", "for", "Beginners")

# iterating over tuple using for loop
for i in tuple_1:
    print(i)
```

```python
# given dictionary
dict_1 = {
    1: "Tpoint",
    2: "Tech"
}

# iterating over dictionary using for loop
for i in dict_1:
    print(i, ":", dict_1[i])
```

```python
# python example to iterate over sequence using index

# given list
fruit_basket = ['apple', 'banana', 'orange', 'mango', 'kiwi']

# iterating over the index of elements in the list
for i in range(len(fruit_basket)):
  # printing the index and the item
  print(i, ":", fruit_basket[i])
```

```python
# else with for loop

for i in range(1, 10):
  print(i)
else:
  print("Loop Finished")
```

```python
count = 0
while count <5 :
    count = count + 1
    print(count," Hi")
```

```python
# Find number divisible by 3 and 11 or 5 and 7
count = 1
while count < 100 :
    if (count%3==0 and count%11==0) :
        print(count," is divisible by 3 and 11")
    if (count%5==0 and count%7==0) :
        print(count," is divisible by 5 and 7")
    count+=1
```
