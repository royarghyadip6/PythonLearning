
#### Example 1: Cosine Similarity Between Two 1D Vectors

* np.dot(A, B) computes the dot product.
* norm(A) and norm(B) compute the magnitude (Euclidean norm).
* The result is a scalar value between -1 and 1 representing similarity.

```python
import numpy as np
from numpy.linalg import norm

A = np.array([2, 1, 2, 3, 2, 9])
B = np.array([3, 4, 2, 4, 5, 5])

# compute cosine similarity
cosine = np.dot(A, B) / (norm(A) * norm(B))

print("Dot :",np.dot(A, B))     # 6 + 4 + 4 + 12 + 10 + 45 = 81
print("Norm(A) :",norm(A))      # 4 + 1 + 2 + 9 + 4 + 81 = 103 > squar root of 103  > 10.14889156509222
print("Norm(B) :",norm(B))      # 9 + 16 + 4 + 16 + 25 + 25 = 95 > square root of 95 > 9.746794344808963
print("(norm(A) * norm(B)) : ",(norm(A) * norm(B))) # 10.14889156509222 * 9.746794344808963 = 98.91915891272024
print("Cosine Similarity:", cosine)         # 81 / 98.91915891272024 = 0.8188504723485274
```

#### Example 2: Cosine Similarity Between a Vector and a Batch of Vectors
* A has 3 vectors (rows), and B is a single vector.
* Cosine similarity is computed row-wise.
* Negative similarity indicates opposite direction.

1. axis=1 > row
2. axis=0 > column

```python
import numpy as np
from numpy.linalg import norm

A = np.array([[2, 1, 2], [3, 2, 9], [-1, 2, -3]])
B = np.array([3, 4, 2])

cosine = np.dot(A, B) / (norm(A, axis=1) * norm(B))

print("np.dot(A, B) :", np.dot(A, B))       # [3*2 + 4*1 + 2*2 , 3*3 + 4*2 + 9*2 , -1*3 + 2*4 + -3*2] = [14 35 -1]
print("norm(A, axis=1) :", norm(A, axis=1)) # [4+1+4 , 9+4+81 , 1+4+9] = [9,94,] > sqrt() > [3, 9.69535971, 3.74165739]
print("norm(B) :", norm(B))     # [9 + 16 + 4] = 29 > sqrt(29) = 5.385164807134504

print("(norm(A, axis=1) * norm(B)) : ",(norm(A, axis=1) * norm(B))) # [5.385164807134504 * 3, 5.385164807134504 * 9.69535971 , 5.385164807134504 * 3.74165739] = [16.15549442 52.21110993 20.14944168]
print("Cosine Similarity:", cosine)     # [14/16.15549442, 35/52.21110993, -1/20.14944168] = [ 0.86657824  0.67035541 -0.04962917]

```

#### Example 3: Cosine Similarity Between Two Matrices (Row-wise)
* Element-wise multiplication is done row-by-row.
* np.sum(..., axis=1) gives the dot products.
* Each output value is similarity between the corresponding rows of A and B.

```python
import numpy as np
from numpy.linalg import norm

A = np.array([[1, 2, 2], [3, 2, 2], [-2, 1, -3]])
B = np.array([[4, 2, 4], [2, -2, 5], [3, 4, -4]])

cosine = np.sum(A * B, axis=1) / (norm(A, axis=1) * norm(B, axis=1))

print("norm(A, axis=1) : ",norm(A, axis=1))
print("norm(B, axis=1) : ",norm(B, axis=1))
print("np.sum(A * B, axis=1) : ",np.sum(A * B, axis=1))
print("(norm(A, axis=1) * norm(B, axis=1)) : ",(norm(A, axis=1) * norm(B, axis=1)))
print("Cosine Similarity:", cosine)
```
