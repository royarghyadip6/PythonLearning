```python
# lambda/annonymus function
def cube(x): return x*x*x   # without lambda
cube_l = lambda x : x*x*x  # with lambda

print(cube(7))
print(cube_l(7))
```

```python
a = lambda x,y,z: x*y*z
print(a(3,1,2))
```

```python
# factorial with lambda and ternary operator
b = lambda x : 1 if x == 1 else x * b(x-1)
print(b(5))
```

```python
# fibonacci sum with lambda and ternary operator
c = lambda x :  0 if x == 0 else 1 if x == 1 else c(x-1) + c(x-2)
print(c(10))
```

```python
# fibonacci series
def abc(x) :
    a,b = 0,1
    for i in range(x) :
        if i == 0 :
            print(0,end=' ')
        elif i == 1 :
            print(1,end=' ')
        else :
            # temp = b
            # b = a+b
            # a = temp
            b = b ^ a
            print(b,end=' ')
abc(10)
```
