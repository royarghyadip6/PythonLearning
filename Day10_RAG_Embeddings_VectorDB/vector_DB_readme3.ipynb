# VECTOR DATABASES NOTES

---

## 1. What is a Vector Database?

A **Vector Database** is a specialized database designed to **store, index, and search vector embeddings efficiently**.

These vectors represent the **semantic meaning** of data such as text, images, audio, or code.

**In simple words:**

> A vector database stores meanings (numbers) and helps find the most similar meanings quickly.

---

## 2. Why Do We Need Vector Databases?

Traditional databases (SQL / NoSQL):

* Work well with exact values
* Use equality or range queries

But they **cannot efficiently search by meaning**.

Vector databases are required to:

1. Perform fast similarity search (nearest neighbors)
2. Scale to millions or billions of embeddings
3. Support semantic search
4. Power RAG systems
5. Reduce LLM hallucinations
6. Enable real-time AI applications

---

## 3. What Problem Do Vector Databases Solve?

### Without Vector DB ❌

* Slow brute-force similarity checks
* High memory usage
* Poor scalability

### With Vector DB ✅

* Optimized indexing (ANN)
* Fast retrieval
* Production-ready AI search

---

## 4. Core Operations in a Vector Database

Every vector database supports:

1. **Insert** – store embeddings
2. **Index** – organize vectors for fast search
3. **Search** – find nearest vectors
4. **Filter** – metadata-based filtering
5. **Delete / Update** – manage data lifecycle

---

## 5. How Vector Databases Work Internally

1. Embeddings are stored as high-dimensional vectors
2. Indexing algorithms group similar vectors
3. Query vector is compared against indexed vectors
4. Closest vectors are returned using similarity metrics

Common metrics:

* Cosine similarity
* Euclidean distance
* Dot product

---

## 6. Role of Vector Databases in RAG Systems

Vector DB is the **retrieval engine** of RAG.

**Flow:**

1. Chunk documents
2. Generate embeddings
3. Store in vector DB
4. Convert query to embedding
5. Retrieve most similar chunks
6. Pass chunks to LLM

Without vector DB → RAG is not scalable.

###FAISS (Vector Database – Local / Open Source)

#### What is FAISS?
FAISS (**Facebook AI Similarity Search**) is an open-source library developed by Meta for fast similarity search on vector embeddings.

#### Why FAISS is important
* Very fast vector search
* No cloud dependency
* Full control over data
* Widely used in research and prototypes

#### When to use FAISS
* Learning vector databases
* Local RAG systems
* Experiments and POCs
* Small to medium datasets

#### Limitation
* Not a full database
* No authentication, persistence, or cloud management


```python
import faiss
import numpy as np

dim = 384
vectors = np.random.random((5, dim)).astype("float32")
query = np.random.random((1, dim)).astype("float32")

index = faiss.IndexFlatL2(dim)
index.add(vectors)

D, I = index.search(query, 3)
print(I, D)
```

### Pinecone (Managed Cloud Vector Database)

#### What is Pinecone?
Pinecone is a fully managed, cloud-native vector database designed for production AI systems.

#### Why Pinecone is important
* No infrastructure management
* Scales to billions of vectors
* Highly reliable and fast
* Ideal for enterprise RAG systems

#### When to use Pinecone
* Production GenAI applications
* Enterprise semantic search
* Large-scale RAG systems

#### Limitation
* Paid service
* Vendor dependency


```python
import pinecone

pinecone.init(api_key="API_KEY", environment="ENV")

index = pinecone.Index("demo-index")

index.upsert([
    ("1", [0.1, 0.2, 0.3]),
    ("2", [0.2, 0.1, 0.4])
])

result = index.query(
    vector=[0.1, 0.2, 0.25],
    top_k=2
)

print(result)
```

### Milvus (Open-Source Distributed Vector DB)

#### What is Milvus?
Milvus is an open-source, distributed vector database built for large-scale similarity search.

#### Why Milvus is important
* Handles massive datasets
* Supports ANN indexing (HNSW, IVF)
* Kubernetes-friendly
* No vendor lock-in

#### When to use Milvus
* Enterprise self-hosted AI systems
* Large-scale semantic search
* When infra control is required

#### Limitation
* Complex setup
* Requires infra management

```python
from pymilvus import connections, Collection

connections.connect("default", host="localhost", port="19530")

collection = Collection("demo_collection")

results = collection.search(
    data=[[0.1, 0.2, 0.3]],
    anns_field="embedding",
    limit=3
)

for hit in results[0]:
    print(hit.id, hit.distance)
```
