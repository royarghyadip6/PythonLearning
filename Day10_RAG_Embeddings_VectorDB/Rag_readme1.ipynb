# RAG SYSTEMS & SEMANTIC SEARCH NOTES

---

## 1. What is a RAG System?

**RAG (Retrieval-Augmented Generation)** is an AI architecture that **retrieves relevant information from external data sources and then generates an answer using a Large Language Model (LLM)**.

**In simple words:**

> RAG = Search first → then generate answer

It combines:

* **Retrieval** (finding correct information)
* **Generation** (forming a natural language answer)

---

## 2. Why Do We Need RAG?

LLMs:

* Do not know your private or latest data
* Can hallucinate (generate incorrect answers)
* Are limited by context size

RAG solves this by:

1. Using external documents
2. Fetching only relevant content
3. Grounding answers in real data

---

## 3. Core Building Blocks of RAG

A RAG system has **5 main components**:

1. Data source (documents, PDFs, DBs)
2. Chunking
3. Embeddings
4. Vector database
5. LLM (answer generation)

---

## 4. Step-by-Step RAG Workflow

### Step 1: Document Ingestion

* Load documents (PDF, text, web pages)

---

### Step 2: Chunking

* Break documents into small, meaningful chunks

**Why needed:**

* LLM token limits
* Better retrieval accuracy

Example:

* 1 large document → 200 small chunks

---

### Step 3: Embedding Generation

* Convert each chunk into a numerical vector (embedding)

**Why needed:**

* Machines understand numbers, not raw text
* Similar meanings → similar vectors

---

### Step 4: Store in Vector Database

* Store embeddings with metadata

Examples:

* Pinecone
* FAISS
* Weaviate
* Milvus

**Why needed:**

* Fast similarity search
* Scales to millions of documents

---

### Step 5: User Query Processing

* User question is converted into an embedding

---

### Step 6: Semantic Retrieval

* Find closest matching document chunks using similarity search

---

### Step 7: Answer Generation

* Retrieved chunks + user query → sent to LLM
* LLM generates a grounded, accurate answer

---

## 5. What is Semantic Search?

**Semantic search** finds results based on **meaning**, not exact words.

**In simple words:**

> Semantic search understands *what you mean*, not just *what you type*.

---

## 6. Why Traditional Search Fails

Keyword Search ❌:

* Exact word matching
* Misses synonyms
* No context understanding

Example:

* Query: "Java backend job"
* Document: "Spring Boot developer role"

Keyword search may fail ❌

---

## 7. How Semantic Search Works

Semantic Search uses:

1. Embeddings
2. Vector similarity

**Process:**

1. Convert documents into embeddings
2. Convert query into embedding
3. Find closest vectors
4. Return meaningfully similar results

---

## 8. Example of Semantic Search

Query:

> "How to reduce database latency?"

Retrieved document may contain:

> "Techniques to improve query performance"

Even without shared keywords, meaning matches ✅

---

## 9. Role of Chunking in Semantic Search & RAG

Chunking ensures:

* Fine-grained retrieval
* No loss of context
* Higher precision

Without chunking:

* Entire documents retrieved
* More noise

---

## 10. Role of Embeddings in RAG & Semantic Search

Embeddings:

* Represent meaning numerically
* Enable similarity measurement
* Power vector databases

**Rule:**

> Distance between vectors = semantic similarity

#### What are vector DB
Vector DB is essential for RAG because it stores data(document,text,images etc.) as numerical vectors(embeddings) that captures their semantic meaning.
* Data Storage : raw data is converted into high dimensional vector or embedding.
* Retrival : The vector DB searches for the vector which is most similar to the query vector. Using algorithm **ANN(Approximate Nearest Neighbour)**.
* Augmentation : The data chunks corresponding to the top-matching vectors are retrieved.
* Generation : The retrieved info is combined with the original query and Sent to the LLM for thr final answer.

#### What is Chunking?
Chunking is the process of breaking a large piece of information (text, data, or tasks) into smaller, manageable parts called chunks. Chunking is used in computer science, AI/ML, databases, NLP, psychology, and learning systems.

* Chunking means breaking large data or text into small, manageable parts.
* It helps AI models handle large documents within context limits.
* Improves search and information retrieval accuracy.
* Used in programming for handling large files, streams, and data batches.
* Example: Large document → small chunks → process one by one.

#### What is embedding?
The embedding is the way to represent text as a vector of numbers.

OpenAI Embeddings: text-embedding-ada-002
HuggingFace models: all-MiniLM-L6-v2
#### Why embedding is required?
Embedding is the process of converting text into vectors. Embedding is required for sementic search, instead of checking actual value (e.g. normal SQL query) the near by value will be retrieved.

----

**_Process of Embedding_:**
* **Input** Piece of a text
* **Tokenization** Split text into tokens
* **Model encoding** Tokens are passed through a neural network that has been trained on huge amount of text
* **Output** A high-dimensional vector (768 or 1536 dimensions)



#### What is KNN?
KNN (K-Nearest Neighbors) is a method to find the K most similar data points to a given data point.
In simple words: KNN means “look at the nearest neighbors and decide based on them.”

#### What Does “K” Mean?
K = number of neighbors you look at
Example:
K = 3 → look at 3 closest neighbors
K = 5 → look at 5 closest neighbors

#### Very Simple Real-Life Example
You move to a new city and want to know: “Is this area safe?” You ask: 5 nearest neighbors If: 4 say “Yes” 1 says “No”
👉 You conclude: Area is safe

That’s KNN thinking.

#### Role of K in Semantic search
K = 3 → top 3 results, K = 5 → top 5 results

#### Important Reality
Exact KNN is slow for large data.
So vector databases use:
* ANN (Approximate Nearest Neighbor)
* Indexes like HNSW, IVF
