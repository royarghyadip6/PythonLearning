# Pandas
- Pandas is an open-source software library for the Python programming language, primarily used for data manipulation and analysis
- It makes it easy to work with structured data – data with rows and columns (like spreadsheets or tables).
- It provides data structures and operations for manipulating numerical tables making it a fundamental tool in the data science ecosystem.
- It supports 2 data structures
    1. Series
    2. DataFrame
- Series :
    - A one-dimensional labeled array capable of holding any data type (integers, strings, floats, Python objects, etc.). 
    - It can be thought of as a single column in a spreadsheet or an SQL table.
- DataFrame : 
    - A two-dimensional labeled data structure with columns of potentially different types. 
    - It can be thought of as a spreadsheet, an SQL table, or a dictionary of Series objects

- Key Functions and Uses :
    - Data Cleaning
    - Data analysis
    - Data Manipulation
    - Loading data
    - Data Visualization
    - Machine Learning


```python
import pandas as pd

# Creating a Series
s = pd.Series([1, 3, 5, None, 6, 8])
print("Series:\n", s)

# With the index argument, you can name your own labels.
s1 = pd.Series([2, 4, 5, 9, 6, 7], index = ["a", "b", "c", "d", "e", "f"])
print("Series with labels:\n", s1)

# You can also use a key/value object, like a dictionary, when creating a Series.
# The keys of the dictionary become the labels.
calories = {"day1": 420, "day2": 380, "day3": 390}
s2 = pd.Series(calories)
print("Series with dictionary:\n", s2)
```

```python
# Creating a DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 35],
        'City': ['New York', 'London', 'Paris']}
df = pd.DataFrame(data, index= ["a", "b", "c"])
print("\nDataFrame:\n", df)
```

### loc and iloc
| **Purpose**                      | **loc**                                                     | **iloc**                                             |
| ----------------------- | ----------------------------------------------------------- | ---------------------------------------------------- |
| **What it uses**        | Labels (row/column names)                               | Index positions (integers)                       |
| **Example**             | `df.loc[0:4, 'Name']` – rows 0 to 4, ‘Name’ column by label | `df.iloc[0:4, 1]` – rows 0 to 3, 2nd column by index |
| **Inclusive/exclusive** | Inclusive of both start and end labels                 | End index is exclusive (like Python slicing)     |
| **When to use**         | When you know the row/column labels                         | When you know the row/column index positions         |

```python
# View a single column
print(df["Pulse"], "\n")

# View multiple columns
print(df[["Duration","Pulse"]], "\n")

# View row by index label 
print(df.loc[0:4,"Pulse"], '\n')

# View row by position 
print(df.iloc[0:4,1])
```

# Cleaning the Data

This includes 
- Handling missing data
- Filling missing data
- Handling Duplicates 
- Removing Outliers
- Fix DataTypes 

### Handling Missing data

```python
# Checking missing values
print(df.isnull().sum())

#  Drop missing data

# Drops rows with missing data (default).
temp_df = df.dropna(axis=0) 
print(temp_df, "\n")

# Drops columns with missing data.
temp_df = df.dropna(axis=1) 
print(temp_df, "\n")

# Drops rows/columns only if all values are missing.
temp_df = df.dropna(how='all') 
print(temp_df, "\n")

# Drops rows where specified columns have missing data.
temp_df = df.dropna(subset=['Duration', 'Pulse']) 
print(temp_df)
```

### Filling Missing data

```python
# Fill with constant value (can be value or text)
df_filled = df.fillna(0.0)
# df_filled = df.fillna("Unknown")
print(df_filled, "\n")

# Fill with mean, median or mode
df_filled = df['Calories'].fillna(df['Calories'].mean())
print(df_filled, "\n")

df_filled = df['Calories'].fillna(df['Calories'].median())   # To reflect the changes in original dataframe, make inplace = True
print(df_filled, "\n")

df_filled = df['Calories'].fillna(df['Calories'].mode()[0])
print(df_filled, "\n")

# Forward fill
print("Forward fill")
df_filled = df.ffill()  # Fill the previous non-null value
print(df_filled, "\n")

# Backward fill   
print("Backward fill") 
df_filled = df.bfill()  # Fill the next non-null value
print(df_filled, "\n")

# Fill with different values for different columns
df_filled = df.fillna({
    "Duration" : df["Duration"].mean(),
    "Calories" : df["Calories"].mode()[0],
    "Pulse" : 85.0
})
print(df_filled)
```

### Handling Duplicates

```python
# Check for duplicates - Returns True for duplicate rows
print(df.duplicated(), "\n")

# Drop all duplicate rows
df_unique = df.drop_duplicates()
print(df_unique.duplicated(), "\n")

# Keep last occurrence 
df_unique = df.drop_duplicates(keep='last')   # by default it keeps the first occurrence
print(df_unique)
```

### Removing Outliers

```python
from scipy import stats 

df_temp = df[(abs(stats.zscore(df['Calories'])) < 3)]  # Keep rows within 3 standard deviations 
# Typically, Z > 3 or Z< -3 is considered an outlier
print(df_temp, "\n")
```

### Fix DataTypes

```python
print(df.dtypes)  # Prints the datatype of each column

# Change the datatype of Maxpulse from int to float
df['Maxpulse'] = df['Maxpulse'].astype(float)
print(df.dtypes)

# Change the datatype of duration to int
df['Duration'] = df['Duration'].fillna(0).astype(int)   # NaN values can't be converted to int directly
print(df.dtypes)
```

### Grouping and Aggregation

```python
# Group by city and get average salary
df_group= df1.groupby('City')['Salary'].mean() 
print(df_group, "\n")

# Multiple Aggregations
df_agg = df1.groupby('City').agg({
    'Salary' : ['mean','max'],
    'Years' : 'median'
})
print(df_agg)
```

### Merging and Joining

```python
employees = pd.DataFrame({
    'ID' : [1, 2, 3],
    'Name' : ['Alice', 'Bob', 'Charlie']
})

salaries = pd.DataFrame({
    'ID' : [1, 2, 4],
    'Salary' : [50000, 60000, 70000]
})

# Merge like SQL join
# Left join
merged_left = pd.merge(employees, salaries, on='ID', how='left')
print(merged_left, "\n")

# Right join
merged_right = pd.merge(employees, salaries, on='ID', how='right')
print(merged_right, "\n")

# Inner join
merged_inner = pd.merge(employees, salaries, on='ID', how='inner')
print(merged_inner, "\n")

# Outer join
merged_outer = pd.merge(employees, salaries, on='ID', how='outer')
print(merged_outer, "\n")
```

### Concatenation

```python
df1 = pd.DataFrame({
    'ID' : [1, 2, 3],
    'Name' : ['Alice', 'Bob', 'Charlie']
})

df2 = pd.DataFrame({
    'ID' : [1, 2, 4],
    'Salary' : [50000, 60000, 70000]
})

# Stack vertically even if columns are different.
df_vert = pd.concat([df1, df2], axis=0)
print(df_vert, "\n")

# Stack Horizontally, side by side, index by index
df_hori = pd.concat([df1, df2], axis=1)
print(df_hori, "\n")
```
