# CHUNKING NOTES

---

## 1. What is Chunking?

**Chunking** is the technique of breaking large data, text, or information into **smaller, manageable, and meaningful units (chunks)** so that they can be processed efficiently without losing context.

**In simple words:**

> Chunking = smartly dividing big information into small parts.

---

## 2. Why Do We Need Chunking?

1. **Memory limits** – Humans, computers, and AI models cannot handle large data at once.
2. **Efficient processing** – Smaller chunks process faster and use less memory.
3. **Accuracy improvement** – Helps maintain focus on relevant information.
4. **Scalability** – Enables systems to work with very large datasets.
5. **Error isolation** – Failure in one chunk doesn’t break the entire system.
6. **LLM constraints** – Language models have token/context limits.
7. **Better search & retrieval** – Easy to find relevant pieces instead of scanning everything.

---

## 3. Where Is Chunking Used?

* Artificial Intelligence & NLP
* Large Language Models (ChatGPT, RAG systems)
* File handling and streaming
* Databases (pagination, batch processing)
* Big data systems (Spark, Hadoop)
* Learning & memory systems

---

## 4. Chunking in AI / LLMs (Important)

Large documents cannot be directly given to LLMs due to token limits.

**Typical Flow:**

1. Split document into chunks
2. Convert chunks into embeddings
3. Store embeddings in a vector database
4. Retrieve relevant chunks on user query
5. Send only selected chunks to the LLM

This process is called **RAG (Retrieval-Augmented Generation)**.

---

## 5. Types of Chunking (With Importance)

### 1️⃣ Fixed-Size Chunking

**What it is:**

* Data is split into chunks of equal size (e.g., 500 tokens each).

**Why it is important:**

* Simple to implement
* Predictable chunk size
* Useful when performance matters more than meaning

**Limitation:**

* Can break sentences or concepts

---

### 2️⃣ Semantic Chunking

**What it is:**

* Data is split based on meaning, topics, or paragraphs.

**Why it is important:**

* Preserves context and intent
* Produces highly accurate AI responses
* Best for knowledge-based systems

**Limitation:**

* Slightly complex to implement

---

### 3️⃣ Sentence-Based Chunking

**What it is:**

* Each sentence or group of sentences becomes a chunk.

**Why it is important:**

* Keeps grammatical correctness
* Useful for chatbots and Q&A systems
* Easy to understand and debug

**Limitation:**

* Context may be too small

---

### 4️⃣ Token-Based Chunking

**What it is:**

* Chunks are created based on token count instead of characters or words.

**Why it is important:**

* Prevents LLM token-limit errors
* Ensures compatibility with AI models
* Most reliable for production GenAI apps

**Limitation:**

* Requires tokenizer logic

---

### 5️⃣ Sliding Window Chunking

**What it is:**

* Chunks overlap with each other (shared context).

**Example:**

* Chunk 1: tokens 1–500
* Chunk 2: tokens 400–900

**Why it is important:**

* Maintains continuity across chunks
* Avoids context loss at boundaries
* Improves answer quality

**Limitation:**

* Uses more memory and storage

---

## 6. Chunk Size Guidelines

* **Text documents:** 300–500 tokens
* **Code:** 100–300 tokens
* **Legal/Medical:** Semantic chunking preferred
* **Chatbots:** Sentence or sliding window chunking

---

## 7. Chunking vs Splitting

| Splitting    | Chunking             |
| ------------ | -------------------- |
| Random       | Purpose-driven       |
| No context   | Context-aware        |
| Simple break | Intelligent division |

---

## 8. Benefits of Chunking (Summary)

* Handles large data safely
* Improves AI accuracy
* Reduces memory usage
* Enables scalable systems
* Improves retrieval & search
* Essential for modern AI applications

---

Reading a book:

This structured reading is **chunking**.

#### Fixed Chunking

```python
from langchain_text_splitters import CharacterTextSplitter, TokenTextSplitter, RecursiveCharacterTextSplitter
from langchain_core.documents import Document

docs = [
    Document(page_content="this is document one. this has several sentences" * 3),
    Document(page_content="This is document tow. this has less sentences " )
]

# Split character wise
characterSplitter = CharacterTextSplitter(
    separator= " ", # default seperator is "\n\n"
    chunk_size=10, # splitting will be near by the chunk size, split can happen
    chunk_overlap=0
)

charChunks = characterSplitter.split_documents(docs)

print("Fixed char Chunks size : ",len(charChunks))

for i, c in  enumerate(charChunks):
    print(f'\nchunk{i} : {c.page_content}')
```

```python
# Split token wise
tokenSplitter = TokenTextSplitter(
    separators=[ " ",""],
    chunk_size=10,
    chunk_overlap=0
)
tokenChunks = tokenSplitter.split_documents(docs)

print("Fixed token Chunks size : ",len(tokenChunks))

for i, c in  enumerate(tokenChunks):
    print(f'\nchunk{i} : {c.page_content}')
```

```python

# Split word wise
wordSplitter = RecursiveCharacterTextSplitter(
    separators=[ " ",""],
    chunk_size=10,
    chunk_overlap=0
)

wordChunks = wordSplitter.split_documents(docs)

print("Fixed wordChunks size : ",len(wordChunks))

for i, c in  enumerate(wordChunks):
    print(f'\nchunk{i} : {c.page_content}')
```

#### Fixed size Chunking with overlap

```python
text = "Sentence one. Sentence two. Sentence three. " * 10

print(text)
doc = Document(page_content=text)

chunk_size=15
chunk_overlap=3  # the character before and after the chunk will be overlap

charSplitter = CharacterTextSplitter(
    separator= "",
    chunk_size=chunk_size,
    chunk_overlap=chunk_overlap
)

chunks = charSplitter.split_documents([doc])

print("Fixed Chunks size : ",len(chunks))

for i, c in  enumerate(chunks):
    print(f'\nchunk{i}:{c.page_content}:{len({c.page_content})}')


for i in range(len(chunks)-1):
    a = chunks[i].page_content
    b = chunks[i+1].page_content

    overlap_a = a[-chunk_overlap:]
    overlap_b = b[:chunk_overlap]

    print(f"a={a} | b={b}")
    print(f"from the end of chunk{i}:{repr(overlap_a)}")
    print(f"from start of chunk{i+1}:{repr(overlap_b)}")
    print(f"Identical ? {repr(overlap_a) == repr(overlap_b)}")
```

#### Sentence based chunking (NLTK/spaCy splitters)

```python
# pip install nltk
"""
Uses punctuation AND spacing patterns

Expects a space after . to detect sentence boundaries reliably
"""
import nltk
nltk.download("punkt")
nltk.download("punkt_tab")

from langchain_text_splitters import NLTKTextSplitter
from langchain_core.documents import Document

text = ("""
Sentence one is small. Sentence two is large. Sentence three is very vey large. Sentence Four is important. Sentence Five. Sentence six. Sentence seven. Uses punctuation AND spacing patterns. Expects a space after to detect sentence boundaries reliably.
""")
doc = Document(page_content=text)
nltk_splitter = NLTKTextSplitter(
    chunk_size=80,
    chunk_overlap=40
)

chunks = nltk_splitter.split_documents([doc])

print("Sentence chunks:",chunks)
print("Fixed Chunks size : ",len(chunks))

for i, c in  enumerate(chunks):
    print(f'\nchunk{i}:{c.page_content}:{len(c.page_content)}')
```

#### Paragraph Based

```python
from langchain_core.documents import Document

text = """
Para 1 : Large Language Models process text as tokens, not characters.

Para 2 : Token-based chunking helps avoid context window overflow.

Para 3 : This is useful when sending long documents to an LLM.
"""

paragraph = [pgph.strip() for pgph in text.split("\n\n") if pgph.strip()]

print("paragraph : ",paragraph)

chunks =  [Document(page_content=para) for para in paragraph]

print("Sentence chunks:",chunks)
print("Fixed Chunks size : ",len(chunks))

for c in chunks:
    print(f'\n{c.page_content}:{len({c.page_content})}')
```

#### Section/Heading based chunking (Markdown/docs)

```python
from langchain_text_splitters import MarkdownHeaderTextSplitter

md_text = """
# Intro
Para 1 : Large Language Models process text as tokens, not characters.

## Explain
Para 2 : Token-based chunking helps avoid context window overflow.

### Summary
Para 3 : This is useful when sending long documents to an LLM.
"""

headerSplit = MarkdownHeaderTextSplitter(
    headers_to_split_on=[
        ("#" , "Header1"),
        ("##" , "Header2"),
        ("###" , "Header3"),
    ]
)

sections = headerSplit.split_text(md_text)

for i, section in enumerate(sections):
    print(f'\nchunk{i}:{section.page_content}:{len(section.page_content)}')
```

#### Semantic Chunking

```python
import nltk
import numpy as np
from nltk.tokenize import sent_tokenize
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Download tokenizer (run once)
nltk.download("punkt")

# -----------------------------
# Long text input
# -----------------------------
text = """
Artificial Intelligence is transforming modern technology. It is widely used in healthcare for disease detection
and drug discovery. Doctors rely on AI-powered tools to analyze medical images accurately.

Machine learning is a subset of AI that focuses on learning from data. Supervised learning requires labeled datasets,
while unsupervised learning works with unlabeled data. Deep learning is a branch of machine learning that uses neural networks.

Climate change is one of the biggest global challenges today. Rising temperatures are causing glaciers to melt.
Governments around the world are investing in renewable energy sources like solar and wind power.

Renewable energy reduces carbon emissions and helps fight global warming. Solar panels and wind turbines are now
more affordable and efficient than ever before.
"""

# -----------------------------
# Step 1: Sentence Tokenization
# -----------------------------
sentences = sent_tokenize(text)

# -----------------------------
# Step 2: Generate Embeddings
# -----------------------------
model = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = model.encode(sentences)

# -----------------------------
# Step 3: Semantic Chunking
# -----------------------------
chunks = []
current_chunk = [sentences[0]]

SIMILARITY_THRESHOLD = 0.65
print("model : ",model)
print("embeddings: ",embeddings)
print("sentences : ",sentences)
for i in range(1, len(sentences)):
    sim = cosine_similarity(
        [embeddings[i - 1]],
        [embeddings[i]]
    )[0][0]

    if sim >= SIMILARITY_THRESHOLD:
        current_chunk.append(sentences[i])
    else:
        chunks.append(" ".join(current_chunk))
        current_chunk = [sentences[i]]

chunks.append(" ".join(current_chunk))

# -----------------------------
# Step 4: Output Chunks
# -----------------------------
for idx, chunk in enumerate(chunks, 1):
    print(f"\n--- Chunk {idx} ---")
    print(chunk)
```

#### Recursive Chunking

```python
import nltk
from nltk.tokenize import sent_tokenize

# Download once
nltk.download("punkt")

# -----------------------------
# Recursive Chunking Function
# -----------------------------
def recursive_chunk_text(
    text,
    max_words=50,
    separators=["\n\n", "\n", ".", " ", ""]
):
    """
    Recursively splits text using decreasingly smaller separators
    until each chunk is <= max_words
    """

    def split_text(text, sep):
        if sep == "":
            return list(text)
        return text.split(sep)

    words = text.split()
    if len(words) <= max_words:
        return [text.strip()]

    for sep in separators:
        parts = split_text(text, sep)
        chunks = []

        current_chunk = ""
        for part in parts:
            candidate = current_chunk + (sep if current_chunk else "") + part
            if len(candidate.split()) <= max_words:
                current_chunk = candidate
            else:
                if current_chunk:
                    chunks.append(current_chunk.strip())
                current_chunk = part

        if current_chunk:
            chunks.append(current_chunk.strip())

        # If split helped, recurse further if needed
        if len(chunks) > 1:
            final_chunks = []
            for chunk in chunks:
                if len(chunk.split()) > max_words:
                    final_chunks.extend(
                        recursive_chunk_text(chunk, max_words, separators[1:])
                    )
                else:
                    final_chunks.append(chunk)
            return final_chunks

    return [text.strip()]

long_text = """
Artificial Intelligence is transforming modern technology.
It is widely used in healthcare for diagnosis and treatment.

Machine learning is a subset of AI that focuses on learning from data.
Deep learning uses neural networks for complex tasks.

Climate change is a major global challenge.
Renewable energy sources like solar and wind are critical.
"""

chunks = recursive_chunk_text(long_text, max_words=20)

for i, chunk in enumerate(chunks, 1):
    print(f"\n--- Chunk {i} ---")
    print(chunk)
```
